```python
import pandas as pd, time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
options = Options()
options.add_argument("--headless=new")
#Extract Tribridges tournament info: Date, time, title, and eventually sections and time control
browser = webdriver.Chrome(options=options)
browser.get('https://tribridgeschessclub.com/events/category/tournaments/')
time.sleep(1)
date_elements = browser.find_elements(By.CLASS_NAME, 'tribe-events-calendar-list__event-datetime-wrapper')
date_texts = []
for elem in date_elements:
    date_texts.append(elem.text)
    time.sleep(1)
link_elements = browser.find_elements(By.CLASS_NAME, 'tribe-events-calendar-list__event-title-link')
link_texts = []
link_hrefs = []
for elem in link_elements:
    link_texts.append(elem.text)
    link_hrefs.append(elem.get_attribute("href"))
cost_texts = []
desc_texts = []
for link in link_hrefs:
    browser.get(link)
    time.sleep(1)
    cost_elements = browser.find_elements(By.CLASS_NAME, 'tribe-events-cost')
    for cost in cost_elements:
        cost_texts.append(cost.text)
    description_elements = browser.find_elements(By.CLASS_NAME, "tribe-events-single-event-description")
    for desc in description_elements:
        desc_texts.append(desc.text)
while len(cost_texts) < len(link_texts):
    cost_texts.append("Not Found")

section_texts = []
time_texts = []
for desc in desc_texts:
    lines = desc.split("\n")
    sections = None
    time_control = None
    for line in lines:
        if "Sections:" in line:
            sections = line.replace("Sections:", "")
            section_texts.append(sections)
        if "Time Control:" in line:
            time_control = line.replace("Time Control:", "")
            time_texts.append(time_control)
    if sections == None:
        section_texts.append("Not Found")
    if time_control == None:
        time_texts.append("Not Found")

df = pd.DataFrame(data = {"Tournament Titles": link_texts, 'Dates': date_texts, "Links": link_hrefs, "Cost": cost_texts, "Description": desc_texts, "Sections": section_texts, "Time Control": time_texts})

for i in range(len(link_texts)):
    print("TITLE: " + df.iloc[i, 0])
    print("LINK: " + df.iloc[i, 2])
    print("DATE AND TIME: " + df.iloc[i, 1])
    print("SECTIONS: " + df.iloc[i, 5])
    print("TIME CONTROL: " + df.iloc[i, 6])
    print("COST: " + df.iloc[i, 3])
    print("DESCRIPTION:\n" + df.iloc[i, 4])
    print("\n")
```
